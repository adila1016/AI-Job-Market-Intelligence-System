```python
# Select the first job
job = analysis_jobs[0]

# Get the original job description
description = job['description']

# Display the first 500 characters
print(description[:500])
```

```python
import re
```

```python
#LOWER CASE
cleaned_text = description.lower()

print(cleaned_text[:500])
```

```python
# ignore .,!()'
cleaned_text = re.sub(r'[^a-zA-Z0-9\s]', '', cleaned_text)
print(cleaned_text[:500])
```

```python
#tokenization
tokens = cleaned_text.split()
print(tokens[:30])
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print(filtered_tokens[:30])
```

```python
## stopword removal
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print(filtered_tokens[:30])
```

```python
## Lematization
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

for word in ["opens", "opportunities", "studies", "cars"]:
    print(word, "→", lemmatizer.lemmatize(word))
```

```python
lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]

print(lemmatized_tokens[:30])
```

```python
lemmatized_text = " ".join(lemmatized_tokens)

print(lemmatized_text[:500])
```

```python
# Save preprocessed text separately
job['lemmatized_description'] = lemmatized_text
```
