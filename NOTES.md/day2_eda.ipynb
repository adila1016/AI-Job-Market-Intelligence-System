```python
ids = [record['id'] for record in jobs]
```

```python
len(ids)
```

```python
id_counts = Counter(ids)
```

```python
id_counts.most_common(5)
```

```python
duplicate_ids = {
    job_id: count
    for job_id, count in id_counts.items()
    if count > 1
}
```

```python
len(duplicate_ids)
```

```python
duplicate_records = [
    record for record in jobs
    if record['id'] in duplicate_ids
]
```

```python
len(duplicate_records)
```

```python
duplicate_records[:2]
```

```python
first_duplicate_id = list(duplicate_ids.keys())[0]

records = [
    record for record in jobs
    if record['id'] == first_duplicate_id
]

len(set(str(record) for record in records))
```

```python
verified_duplicates = 0

for job_id in duplicate_ids:
    records = [
        record for record in jobs
        if record['id'] == job_id
    ]

    if len(set(str(record) for record in records)) == 1:
        verified_duplicates += 1

verified_duplicates
```

```python
with open('datasets/cleaned_jobs.json', encoding='utf-8') as file:
    saved_jobs = json.load(file)

len(saved_jobs)
```

```python
missing_values = {}

for column in saved_jobs[0].keys():
    missing_values[column] = sum(
        1 for job in saved_jobs
        if job[column] is None
    )

missing_values
```

```python
total_records = len(saved_jobs)

missing_percentage = {}

for column, count in missing_values.items():
    percentage = (count / total_records) * 100
    missing_percentage[column] = round(percentage, 2)

missing_percentage
```

```python
for column, percentage in missing_percentage.items():
    if percentage > 0:
        print(column, ":", percentage, "%")
```
